# Brief technological recap

Solar panels are devices that convert solar electromagnetic emission into electric current. They are also called photovoltaic (PV) panels to distinguish them from solar thermal panels, which convert sunlight into heat. I will focus on PV panels so I will use the term solar panels and PV panels interchangeably. 

Solar panels produce electricity in the form of direct current, at a relatively low voltage (12 to 48 V). However, the current that we conventionally use in our homes is alternated current at a much higher voltage of 240 V. This means that to couple solar panels with the residential electric circuit we need to convert the direct, low voltage current into alternated, high voltage current through the use of an inverter.

Nowadays, PV systems are also coupled with electric storage in the form of batteries. As traditional cellphone batteries, electric storage converts the electric energy into chemical energy, which is released when needed. Electric storage is particularly useful in combination with PV panels because it allows to spread the consumption of renewable energy throughout the day. 

In terms of measurements, the power produced by the solar panels is defined as the amount of sunlight converted into electricity every second, measured in Watts (W). The producers of the solar panels list its peak power, which is the electrical power generated when sunlight perpendicular to the solar panel hits it with an irradiance of 1000 W/m2. This reference irradiance mimics the average amount of sunlight hitting the Earth.

The electrical energy produced by a solar panel is obtained by summing the electrical power across a specific time interval. For example, a panel generating 1000 W for an hour will have generated a 1000 Watt hour (Wh) or 1 kWh. Thus, the W is a measure of power (energy per unit of time) while the kWh is a measure of energy. 

Batteries store electrical energy, and their capacity is also generally listed in kWh. When they release their energy, they will do that with a specific power output that is measured in W.

To give an analogy, the capacity of a water reservoir is measured in liters, while the flow is measured in liters per second. The capacity is analogous to the energy while the flow is analogous to the power.

The energy meter connected to our home's electrical panel measures the amount of energy consumed in kWh. Traditional analogical meters cannot be read remotely. Every so often the amount of energy consumed must be reported to the electricity provider to calculate the amount of money that needs to be balanced. Recently, Sibelga, Brussels distribution system operator, has started replacing the old meters with new smart meters that can be read remotely and track the energy consumption every 15 minutes. 

Smart meters are a technological requirement to go from the conventional energy system to a new energy system in which individuals can be at the same time producers and consumers (prosumers) of electricity. It would be impossible for example to coordinate the sharing of electricity produced by rooftop solar across different apartments or even buildings. 

Smart meters also represent a huge opportunity from a research point of view, because of the quantity of data that can be gathered and analyzed on the consumption patterns. This data can be employed to develop more efficient methods for energy allocation, for the coordination of different energy resources needed to cover the energy requirements, or for the prediction of different scenarios of energy procurements. 

As always with data collection, they also pose quite an important concern from a privacy point of view, and robust safeguides have to be put in place to avoid exposing sensitive information.
