# Permitting

## Plug and play balcony solar
The installation itself does not require any certification from Sibelga. The solar panels, inverter and eventual batteries need only to be installed on a circuit that can handle the power production. A regular circuit connected to a 16A breaker should be capable of handling the maximum output power of 800 W without issues. Before installation, it is necessary to request the replacement of the traditional meter with a smart meter from Sibelga. Once the system is installed, it is mandatory to communicate the installation to Sibelga.  

On top of these requirements, any modification of the appearance of the facade generally requires a permit from the urbanism office. On this point, Homegrade explains what are the situations in which solar panels are exempt from permitting. My understanding is that in my case I would need to require a permit from the commune in order to install the solar panels on my balcony.
It also needs to be approved by the assembly of the owners if you live in a shared building, which is most of the people in Brussels.

All in all, the installation of solar panels on the balcony represents a fairly straightforward way to both reduce the individual energy costs and the carbon footprint. However, the technical constraint of a maximum power of 800 W plus the architectural requirements will heavily limit the deployment of this technology. Moreover, the fact that plug and play devices are not eligible for green certificates signals that the Brussels authorities do not believe that this technology can have a significant impact on the energy production and carbon emissions of the grid. From the point of view of the grid operator, small distributed energy resources (DER) are considered a bit of a nuisance, because they can be hard to integrate and can compromise the stability of the grid. 
However, this problem can be greatly mitigated by coupling the installation with battery storage, which limits the amount of energy injected in the grid.
