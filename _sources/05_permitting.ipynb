## Permitting

### Plug and play balcony solar

The installation itself does not require any certification from Sibelga. The solar panels, inverter and eventual batteries need only to be installed on a circuit that can handle the power production. A regular circuit connected to a 16A breaker should be capable of handling the maximum output power of 800 W. Before installation, it is necessary to request the replacement of the traditional meter with a smart meter from Sibelga. Once the system is installed, it is mandatory to communicate the installation to [Sibelga](https://www.sibelga.be/en/faq/renewable-energy/what-do-i-need-to-consider-when-installing-plug-play-solar-panels). 

On top of these requirements, any modification of the appearance of the facade generally requires a permit from the urbanism office. On this point, [Homegrade](https://homegrade.brussels/conseils/urbanisme/travaux-sans-permis-urbanisme-architecte-biens-non-proteges/) explains what are the situations in which solar panels are exempt from permitting. With high likelihood, it also must be approved by the assembly of the owners if you live in a shared building.

All in all, the installation of solar panels on the balcony represents a fairly straightforward way to both reduce the individual energy costs and the carbon footprint. However, the technical constraint of a maximum power of 800 W plus the architectural requirements will heavily limit the deployment of this technology. Moreover, the fact that plug and play devices are not eligible for green certificates signals that the Brussels authorities do not believe that this technology can have a significant impact on the energy production and carbon emissions of the grid. From the point of view of the grid operator, small DERs are considered a bit of a nuisance, because they can be hard to integrate and can compromise the stability of the grid. However, this problem can be greatly mitigated by coupling the installation with battery storage, which limits the amount of energy injected in the grid. 

### Rooftop solar

From a technical point of view, the installation of rooftop solar will have to be conducted by a certified company which will intall the solar panels and connect them to the building electrical panel. Sibelga provides a [guide](https://www.sibelga.be/en/connections-meters/renewable-energy/solar-panels/produce-my-own-energy-in-10-steps) in ten steps to follow before producing green energy: 
* if the project is over 10 kVA (this is the rating of the inverter, which is generally a bit smaller than the peak generation capacity of the solar panels), you must ask for an agreement in principle from Sibelga,
* request a smart meter,
* install the solar panels,
* request the inspection,
* receive the Sibelga certificate,
* Certify your green meter,
* Obtain access to the Sibelga's green meter application, and Brugel's extranet for green certificates.
* Find your green certificates on Brugel's extranet,
* Sell your green certificates.

Regarding the urbanistic permits, [Homegrade](https://homegrade.brussels/conseils/urbanisme/travaux-sans-permis-urbanisme-architecte-biens-non-proteges/) explains that the installation of rooftop solar generally does not require a permit if the solar panels are not visible form the street, and if the building is not listed or less than 20 meters from a listed building.
